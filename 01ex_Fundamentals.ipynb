3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
def Hello(func):
    def wrapper(*args, **kwargs):
        print("Hello!")
        return func(*args, **kwargs)
    return wrapper

@Hello
def greet(name):
    return f"Welcome, {name}!"

@Hello
def sqr(x):
    return x**(1/2)

@Hello
def cube(x):
    return x**3 

@Hello
def power(x):
    x = sqr(cube(x**4))
    return x


rizz = greet("Alice")
print(rizz)
number = int(input("Enter a number: "))
print("The 6th power of the number is: ", power(number))
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
def factorial1(n):
    if n == 0:
        return 1
    else:
        return n * factorial1(n - 1)


def factorial2(n):
    a= 1
    for i in range(1, n + 1):
        a *= i
    return a


print([factorial1(i) for i in range(10)])
print([factorial2(i) for i in range(10)])   
````

````python
%timeit factorial1(1000)
%timeit factorial2(1000)
````

6\. Edit the class defintion to add an instance attribute of is_hungry = True to the Dog class. Then add a method called eat() which changes the value of is_hungry to False when called. Figure out the best way to feed each dog and then output “My dogs are hungry.” if all are hungry or “My dogs are not hungry.” if all are not hungry. The final output should look like this:

`I have 3 dogs. 
Tom is 6. 
Fletcher is 7. 
Larry is 9. 
And they're all mammals, of course. 
My dogs are not hungry.
`

````python
# Parent class
class Dog:

    # Class attribute
    species = 'mammal'

    # Initializer / Instance attributes
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.is_hungry= True

    # instance method
    def description(self):
        return "{} is {} years old".format(self.name, self.age)

    # instance method
    def speak(self, sound):
        return "{} says {}".format(self.name, sound)
    
    # is hungry
    def eat(self):
        self.is_hungry = False

# Child class (inherits from Dog class)
class RussellTerrier(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)

# Child class (inherits from Dog class)
class Bulldog(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)

my_dogs = [Bulldog("Tom", 6), RussellTerrier("Fletcher", 7), Dog("Larry", 9)]

print(f"I have {len(my_dogs)} dogs.")
for dog in my_dogs:
    print(f"{dog.name} is {dog.age}.")
print("And they're all mammals, of course.")

for dog in my_dogs:
    dog.eat()

if all(dog.is_hungry for dog in my_dogs):
    print("My dogs are hungry.")
elif not any(dog.is_hungry for dog in my_dogs):
    print("My dogs are not hungry.")
    
````
